```python
class Employee:
    def __init__(self, first, last, salary):
        self.first = first
        self.last = last
        self.salary = salary
        self.email = '{}.{}@gmail.com'.format(first, last)

    def fullname(self):
        return '{} {}'.format(self.first, self.last)

    def increment(self, amount):
        self.salary += amount

    def __repr__(self):
        return "Employee('{}', '{}', {})".format(self.first, self.last, self.salary)
```

```python
emp1 = Employee('Vishwajeet', 'Sonkhedkar', 100000)
print(emp1.increment(1000))
```
